```python
from functools import reduce
```

```python
def add(x,y):
    return x+y

def double(x):
    return 2*x

def square(x):
    return x**2
```

```python
def create_composite2(*functions):
    def composed(*args, **kwargs):
        result = args  # Start with the initial arguments
        for func in reversed(functions):  # Apply each function in reverse order
            result = (func(*result, **kwargs),)  # Call function with unpacked arguments
        return result[0]  # Return the final result
    return composed
```

```python
f2 = create_composite2(square,add)
f3 = create_composite2(square,double)
```

```python
f2(3,2)
```

```python
f3(4)
```
